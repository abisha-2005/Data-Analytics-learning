PERMUTATION

```python
from itertools import permutations
count= 0
data= [1,2,3]
perm= permutations(data,2)
for i in perm:
  print(i)
  count+=1
print("number of permutation: ",count)
```

COMBINATION WITH LENGTH 2:

```python
from itertools import combinations
count=0
data=[1,2,3]
combi= combinations(data,2)
for i in combi:
  print(i)
  count+=1
print("number of combinations: ",count)
```

```python
from itertools import combinations_with_replacement as cr
count=0
data=[1,2,3]
data2=[2,3,1]
comb=cr(data,2)
for i in comb:
  print(i)
  count+=1
print("number of combination with replacement: ",count)#it is used when we need 1,1->2,2->3,3
```
